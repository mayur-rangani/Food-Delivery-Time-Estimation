# 01. Data Collection and Cleaning

This notebook loads the food delivery dataset and performs the cleaning steps from the original `models.ipynb`.

**Dataset path:** `../dataset/foodtimedataset.csv`

## 1. Import libraries

```python
import pandas as pd
import numpy as np
```

## 2. Load the dataset

```python
df = pd.read_csv("../dataset/foodtimedataset.csv")

df["Vehicle Type"] = df["Vehicle Type"].str.strip().str.title()

print(df["Vehicle Type"].unique())

print("Original shape:", df.shape)
df.head()
```

## 3. Clean the dataset

- Remove spaces from column names and text values.
- Convert `Delivery Time` to numeric.
- Remove rows where delivery time is missing or not positive.

```python
df.columns = df.columns.str.strip()

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

df["Delivery Time"] = pd.to_numeric(df["Delivery Time"], errors="coerce")
df = df.dropna(subset=["Delivery Time"])
df = df[df["Delivery Time"] > 0]

print("Cleaned shape:", df.shape)
df.head()
```

## 4. Check missing values and duplicates

```python
print("Missing values by column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 5. Save the cleaned dataset (optional)

```python
import os

os.makedirs("../dataset", exist_ok=True)
df.to_csv("../dataset/cleaned_foodtimedataset.csv", index=False)
print("Cleaned dataset saved to dataset/cleaned_foodtimedataset.csv")
```
